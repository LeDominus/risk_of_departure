```python
churn_data_new = pd.read_csv('../data/raw/churn_data_new.csv')

churn_data_new["TotalCharges"] = churn_data_new["TotalCharges"].replace(' ', np.nan)
churn_data_new["TotalCharges"] = pd.to_numeric(churn_data_new["TotalCharges"], errors='coerce')
churn_data_new["TotalCharges"] = churn_data_new["TotalCharges"].fillna(0)
churn_data_new["TotalCharges"] = churn_data_new["TotalCharges"].astype(float)

churn_data_new.to_csv('../data/raw/churn_data_new.csv')
churn_data_new.info()
```
