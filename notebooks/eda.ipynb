```python
print(f"""
    ОПИСАНИЕ ФИЧЕЙ:
1. CustomerID - уникальный идентификатор клиента
2. gender - пол клиента (Male/Female)
3. SeniorCitizen - является ли клиент пенсионером (1 - да, 0 - нет)
4. Partner - наличие партнёра/супруга (Yes/No)
5. Dependents - наличие иждивенцев (Yes/No)
6. tenure - срок обслуживания клиента в месяцах
7. PhoneService - наличие телефонной связи (Yes/No)
8. MultipleLines - наличие нескольких телефонных линий (Yes/No/No phone service)
9. InternetService - тип интернет-услуги (DSL/Fiber optic/No)
10. OnlineSecurity - подключена ли услуга онлайн-безопасности (Yes/No/No internet service)
11. OnlineBackup - подключено ли онлайн-резервное копирование (Yes/No/No internet service)
12. DeviceProtection - подключена ли защита устройства (Yes/No/No internet service)
13. TechSupport - наличие технической поддержки (Yes/No/No internet service)
14. StreamingTV - подключено ли потоковое телевидение (Yes/No/No internet service)
15. StreamingMovies - подключены ли потоковые фильмы (Yes/No/No internet service)
16. Contract - тип контракта (Month-to-month, One year, Two year)
17. PaperlessBilling - использование безбумажных счетов (Yes/No)
18. PaymentMethod - способ оплаты (Electronic check, Mailed check, Bank transfer, Credit card)
19. MonthlyCharges - ежемесячная плата за услуги
20. TotalCharges - общая сумма оплаты за весь период (может быть строкой, необходимо преобразование в число)
21. Churn - целевая переменная: отток клиента (Yes/No)
""")

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Создание фигуры с 2x3 подграфиками
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
fig.suptitle('Exploratory Data Analysis - Telco Customer Churn', fontsize=16)

# Общая настройка сетки для всех осей
for ax in axes.flat:
    ax.grid(True, which='both', linestyle='--', alpha=0.5)

# 1. Гистограмма распределения MonthlyCharges
sns.histplot(df['MonthlyCharges'], bins=30, kde=True, ax=axes[0,0], color='skyblue')
axes[0,0].set_title('Distribution of Monthly Charges')
axes[0,0].set_xlabel('Monthly Charges')

# 2. Гистограмма распределения tenure
sns.histplot(df['tenure'], bins=30, kde=True, ax=axes[0,1], color='salmon')
axes[0,1].set_title('Distribution of Tenure')
axes[0,1].set_xlabel('Tenure (months)')

# 3. Круговая диаграмма Churn
churn_counts = df['Churn'].value_counts()
axes[0,2].pie(churn_counts, labels=['No', 'Yes'], autopct='%1.1f%%', 
              startangle=90, colors=['lightgreen', 'lightcoral'])
axes[0,2].set_title('Churn Distribution')

# 4. Круговая диаграмма Contract
contract_counts = df['Contract'].value_counts()
axes[1,0].pie(contract_counts, labels=contract_counts.index, autopct='%1.1f%%',
              startangle=90, colors=sns.color_palette('pastel'))
axes[1,0].set_title('Contract Type Distribution')

# 5. Линейный график: средний MonthlyCharges по tenure
avg_monthly_by_tenure = df.groupby('tenure')['MonthlyCharges'].mean().reset_index()
sns.lineplot(data=avg_monthly_by_tenure, x='tenure', y='MonthlyCharges', 
             ax=axes[1,1], marker='o', color='purple')
axes[1,1].set_title('Average Monthly Charges by Tenure')
axes[1,1].set_xlabel('Tenure')
axes[1,1].set_ylabel('Avg Monthly Charges')

# 6. Линейный график: средний TotalCharges по tenure
avg_total_by_tenure = df.groupby('tenure')['TotalCharges'].mean().reset_index()
sns.lineplot(data=avg_total_by_tenure, x='tenure', y='TotalCharges', 
             ax=axes[1,2], marker='s', color='teal')
axes[1,2].set_title('Average Total Charges by Tenure')
axes[1,2].set_xlabel('Tenure')
axes[1,2].set_ylabel('Avg Total Charges')

plt.tight_layout()
plt.savefig("../data/result/eda_graphic.png")
plt.show()
```

## Feature Engineering

```python
df['MonthlyCharges'] = np.log1p(df['MonthlyCharges'])
df['tenure'] = np.log1p(df['tenure'])

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
fig.suptitle('Exploratory Data Analysis - Telco Customer Churn', fontsize=16)

# Общая настройка сетки для всех осей
for ax in axes.flat:
    ax.grid(True, which='both', linestyle='--', alpha=0.5)

# 1. Гистограмма распределения MonthlyCharges
sns.histplot(df['MonthlyCharges'], bins=30, kde=True, ax=axes[0,0], color='skyblue')
axes[0,0].set_title('Distribution of Monthly Charges')
axes[0,0].set_xlabel('Monthly Charges')

# 2. Гистограмма распределения tenure
sns.histplot(df['tenure'], bins=30, kde=True, ax=axes[0,1], color='salmon')
axes[0,1].set_title('Distribution of Tenure')
axes[0,1].set_xlabel('Tenure (months)')
```

```python
from sklearn.preprocessing import LabelEncoder

cat_features = list(df.drop(['tenure', "MonthlyCharges", "SeniorCitizen", "TotalCharges", "Churn"], axis=1).columns)

df_lr = df.copy()

le = LabelEncoder()
for feature in cat_features:
    print(f"Число уникальных значений в столбце: {feature} - {df_lr[feature].nunique()}")
    df_lr[feature] = le.fit_transform(df_lr[feature])
```

```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

X_lr = df_lr.drop(["Churn", "customerID"], axis=1)
y_lr = df_lr["Churn"]

X_train_lr, X_test_lr, y_train_lr, y_test_lr = train_test_split(X_lr, y_lr, test_size=0.15, random_state=RANDOM_STATE)
print(X_train_lr.shape, X_test_lr.shape, y_train_lr.shape, y_test_lr.shape)
```

```python
type(y_lr)
```

## Fit model and predict

### LogisticRegression model

```python
import warnings
from sklearn.linear_model import LogisticRegression

warnings.filterwarnings("ignore")

lr = LogisticRegression(
    n_jobs=-1,
    verbose=100,
    max_iter=100,
    class_weight="balanced"
    )
lr.fit(X_train_lr, y_train_lr)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve, confusion_matrix, classification_report

y_pred = lr.predict(X_test_lr)
y_pred_proba = lr.predict_proba(X_test_lr)[:, 1]

print("Accuracy:", np.round(accuracy_score(y_test_lr, y_pred), 4))
print("Precision:", np.round(precision_score(y_test_lr, y_pred), 4))
print("Recall:", np.round(recall_score(y_test_lr, y_pred), 4))
print("F1:", np.round(f1_score(y_test_lr, y_pred), 4))
print("ROC-AUC:", np.round(roc_auc_score(y_test_lr, y_pred_proba), 4))
print("\nClassification Report:\n", classification_report(y_test_lr, y_pred))

fpr, tpr, thresholds = roc_curve(y_test_lr, y_pred_proba)
plt.figure(figsize=(14, 8))
plt.plot(fpr, tpr, label='ROC curve (area = %0.2f)' % roc_auc_score(y_test_lr, y_pred_proba))
plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc="lower right")
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.show()

cm = confusion_matrix(y_test_lr, y_pred)
plt.figure(figsize=(14, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
import os
from skl2onnx import __max_supported_opset__
from skl2onnx import to_onnx
from sklearn.linear_model import LogisticRegression

# 1. Ваше обучение модели
lr = LogisticRegression(
    n_jobs=-1,
    verbose=100,
    max_iter=100,
    class_weight="balanced"
)
lr.fit(X_train_lr, y_train_lr)

# 2. Создаем целевую директорию, если она еще не существует
output_dir = "../data/result"
os.makedirs(output_dir, exist_ok=True)
onnx_path = os.path.join(output_dir, "logistic_regression.onnx")

# 3. Конвертация в формат ONNX
# Передаем пример данных X_train_lr, чтобы библиотека автоматически 
# определила тип (например, float32) и количество признаков (shape).
sample_input = X_train_lr[:1].to_numpy().astype(np.float32)

onx = to_onnx(
    lr, 
    sample_input, 
    target_opset=__max_supported_opset__
)

# 4. Сохранение файла на диск
with open(onnx_path, "wb") as f:
    f.write(onx.SerializeToString())

print(f"\nМодель успешно сохранена по пути: {onnx_path}")
```

```python
feature_names = X_train_lr.columns
coefs = lr.coef_[0]
importance_df = pd.DataFrame({
    'feature': feature_names,
    'coefficient': coefs,
    'abs_coefficient': np.abs(coefs)
}).sort_values('abs_coefficient', ascending=False)

plt.figure(figsize=(14, 8))
sns.barplot(x='abs_coefficient', y='feature', data=importance_df.head(20), palette='RdBu_r')
plt.title('Top 20 Feature Coefficients (Logistic Regression)')
plt.xlabel('Coefficient value')
plt.ylabel('Feature')
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("../data/result/feature_importances_lr.png")
plt.show()
```

### LightGBM model

```python
X = df.drop(["Churn", "customerID"], axis=1)
y = df["Churn"]

cat_features = list(X.select_dtypes(include=[str]))
for feature in cat_features:
    X[feature] = X[feature].astype("category")

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=RANDOM_STATE)
```

```python
X.info()
```

```python
import lightgbm as lgb

lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.1,
    num_leaves=31,
    max_depth=-1,
    random_state=42,
    n_jobs=-1,
    verbose=-1,
    reg_alpha=1,
    reg_lambda=1
)

lgb_model.fit(X_train, y_train)
```

```python
y_pred_lgb = lgb_model.predict(X_test)
y_pred_proba_lgb = lgb_model.predict_proba(X_test)[:, 1]

print("Accuracy:", np.round(accuracy_score(y_test, y_pred_lgb), 4))
print("Precision:", np.round(precision_score(y_test, y_pred_lgb), 4))
print("Recall:", np.round(recall_score(y_test, y_pred_lgb), 4))
print("F1:", np.round(f1_score(y_test, y_pred_lgb), 4))
print("ROC-AUC:", np.round(roc_auc_score(y_test, y_pred_proba_lgb), 4))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lgb))

fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba_lgb)
plt.figure(figsize=(14, 8))
plt.plot(fpr, tpr, label='ROC curve (area = %0.2f)' % roc_auc_score(y_test, y_pred_proba_lgb))
plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc="lower right")
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.show()

cm = confusion_matrix(y_test, y_pred_lgb)
plt.figure(figsize=(14, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
importance_df = pd.DataFrame({
    'feature': lgb_model.feature_name_,
    'importance': lgb_model.feature_importances_
}).sort_values('importance', ascending=False).head(20)

plt.figure(figsize=(10, 8))
sns.barplot(x='importance', y='feature', data=importance_df, palette='viridis')
plt.title('Feature Importance - LightGBM')
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("../data/result/feature_importances_lgb.png", bbox_inches='tight')
plt.show()
```
